```python
# Specify the full volume path
target_path = "/Volumes/linkedin_scraper_etl/live_linkedin/linkedin_raw_json/"

def delete_all(path):
    items = dbutils.fs.ls(path)
    for item in items:
        if item.isDir():
            delete_all(item.path)
            dbutils.fs.rm(item.path)
        else:
            dbutils.fs.rm(item.path)

delete_all(target_path)
dbutils.fs.rm(target_path)
```
